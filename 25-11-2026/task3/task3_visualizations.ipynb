# Employee Attrition Data Visualization

## Objective
Analyze employee data using five different visualizations to identify salary patterns and employee age distribution.

## Visualizations
The following charts are used:
1. Bar Chart – Average salary by department
2. Histogram – Employee age distribution
3. Scatter Plot – Age vs Monthly Income
4. Line Chart – Average salary by job level
5. Box Plot – Monthly income by department

```python

chart_insights = {
    "01_bar_salary_by_department.png":
        "Sales department has the highest average monthly salary among the three departments.",

    "02_histogram_age_distribution.png":
        "The histogram shows the distribution of employees across different age groups.",

    "03_scatter_age_vs_income.png":
        "Monthly income varies among employees across different age groups.",

    "04_line_salary_by_job_level.png":
        "Average monthly salary increases as job level increases.",

    "05_boxplot_income_by_department.png":
        "The box plot compares salary distribution and outliers across departments."
}

for chart, insight in chart_insights.items():
    print("Chart:", chart)
    display(Image(filename=str(FIGURES_DIR / chart)))
    print("Insight:", insight)
```
